```python
 !pip install -q tensorflow scikit-learn matplotlib opencv-python-headless

import os
import time
import zipfile
import urllib.request
import numpy as np
import cv2
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec

import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers, regularizers
from tensorflow.keras.applications import ResNet50, EfficientNetB0
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.utils import to_categorical

from sklearn.model_selection import train_test_split
from sklearn.neural_network import MLPClassifier
from sklearn.metrics import accuracy_score
from sklearn.preprocessing import LabelEncoder

print("TensorFlow version:", tf.__version__)
print("GPU available:", tf.config.list_physical_devices('GPU'))


# ────────────────────────────────────────────────────────────
# CELL 2 — Download & Extract GTSRB Dataset
# ────────────────────────────────────────────────────────────

# We use the Kaggle-hosted or direct GTSRB training subset.
# For Colab: run the Kaggle API setup OR use the mirror below.

# --- Option A: Kaggle API (uncomment if you have kaggle.json) ---
# from google.colab import files
# files.upload()  # upload kaggle.json
# !mkdir -p ~/.kaggle && mv kaggle.json ~/.kaggle/ && chmod 600 ~/.kaggle/kaggle.json
# !kaggle datasets download -d meowmeowmeowmeowmeow/gtsrb-german-traffic-sign
# !unzip -q gtsrb-german-traffic-sign.zip -d gtsrb

# --- Option B: Direct download mirror (no Kaggle account needed) ---
DATASET_URL = "https://sid.erda.dk/public/archives/daaeac0d7ce1152aea9b61d9f1e19370/GTSRB_Final_Training_Images.zip"
DATASET_ZIP = "GTSRB_Training.zip"
DATASET_DIR = "gtsrb"

if not os.path.exists(DATASET_DIR):
    print("Downloading GTSRB dataset (~260 MB) …")
    urllib.request.urlretrieve(DATASET_URL, DATASET_ZIP)
    print("Extracting …")
    with zipfile.ZipFile(DATASET_ZIP, 'r') as zf:
        zf.extractall(DATASET_DIR)
    print("Done.")
else:
    print("Dataset already present.")
```

```python
# ────────────────────────────────────────────────────────────
# CELL 3 — Load & Preview Raw Images
# ────────────────────────────────────────────────────────────

IMG_SIZE   = 32          # resize all images to 32×32
MAX_SAMPLES = 500        # samples PER CLASS (set None for full dataset)
N_CLASSES  = 43          # GTSRB has 43 traffic sign classes

def load_gtsrb(base_dir, img_size=32, max_per_class=None):
    """Walk GTSRB folder structure, load & resize images."""
    train_path = os.path.join(base_dir, "GTSRB", "Final_Training", "Images")
    if not os.path.isdir(train_path):
        # Fallback: some mirrors unzip differently
        for root, dirs, files in os.walk(base_dir):
            if "00000" in dirs:
                train_path = root
                break
    images, labels = [], []
    for class_id in range(N_CLASSES):
        class_dir = os.path.join(train_path, f"{class_id:05d}")
        if not os.path.isdir(class_dir):
            continue
        files = [f for f in os.listdir(class_dir) if f.endswith(".ppm")]
        if max_per_class:
            files = files[:max_per_class]
        for fname in files:
            img = cv2.imread(os.path.join(class_dir, fname))
            if img is None:
                continue
            img = cv2.resize(img, (img_size, img_size))
            images.append(img)
            labels.append(class_id)
    return np.array(images), np.array(labels)

X_raw, y = load_gtsrb(DATASET_DIR, IMG_SIZE, MAX_SAMPLES)
print(f"Loaded {len(X_raw)} images | Shape: {X_raw.shape} | Classes: {N_CLASSES}")

# Quick preview — 10 random samples
fig, axes = plt.subplots(2, 5, figsize=(14, 6))
fig.suptitle("GTSRB — Raw Sample Images", fontsize=14, fontweight='bold')
for ax in axes.flat:
    idx = np.random.randint(len(X_raw))
    ax.imshow(cv2.cvtColor(X_raw[idx], cv2.COLOR_BGR2RGB))
    ax.set_title(f"Class {y[idx]}", fontsize=9)
    ax.axis('off')
plt.tight_layout()
plt.savefig("preview_raw.png", dpi=100)
plt.show()
```

```python
# ────────────────────────────────────────────────────────────
# CELL 4 — Classical Preprocessing
#   • BGR → HSV / Grayscale conversion
#   • Contrast enhancement (CLAHE)
#   • Image normalization
# ────────────────────────────────────────────────────────────

def classical_preprocess(img_bgr):
    """Full classical preprocessing pipeline."""
    # 1. Color space conversions
    img_hsv   = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2HSV)
    img_gray  = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2GRAY)
    img_rgb   = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)

    # 2. CLAHE contrast enhancement on the V channel of HSV
    clahe = cv2.createCLAHE(clipLimit=2.0, tileGridSize=(4, 4))
    h, s, v = cv2.split(img_hsv)
    v_eq = clahe.apply(v)
    img_hsv_eq = cv2.merge([h, s, v_eq])
    img_enhanced = cv2.cvtColor(img_hsv_eq, cv2.COLOR_HSV2BGR)

    # 3. Normalize to [0, 1]
    img_norm = img_enhanced.astype(np.float32) / 255.0

    return img_gray, img_hsv, img_enhanced, img_norm

# Visualise pipeline on one sample
sample_bgr = X_raw[42]
gray, hsv, enhanced, norm = classical_preprocess(sample_bgr)

fig, axes = plt.subplots(1, 5, figsize=(16, 4))
fig.suptitle("Classical Preprocessing Pipeline", fontsize=13, fontweight='bold')
titles = ["Original (RGB)", "Grayscale", "HSV", "CLAHE Enhanced", "Normalised"]
imgs   = [
    cv2.cvtColor(sample_bgr, cv2.COLOR_BGR2RGB),
    gray, cv2.cvtColor(hsv, cv2.COLOR_HSV2RGB),
    cv2.cvtColor(enhanced, cv2.COLOR_BGR2RGB),
    norm
]
cmaps = [None, 'gray', None, None, None]
for ax, title, img, cmap in zip(axes, titles, imgs, cmaps):
    ax.imshow(img, cmap=cmap)
    ax.set_title(title, fontsize=10)
    ax.axis('off')
plt.tight_layout()
plt.savefig("preprocessing_pipeline.png", dpi=100)
plt.show()

```

```python
# ────────────────────────────────────────────────────────────
# CELL 5 — Edge Detection
#   • Manual Prewitt & Laplacian kernels (NumPy)
#   • cv2.Sobel comparison
# ────────────────────────────────────────────────────────────

def manual_prewitt(gray_img):
    """Prewitt operator implemented with raw NumPy convolution."""
    kx = np.array([[-1, 0, 1], [-1, 0, 1], [-1, 0, 1]], dtype=np.float32)
    ky = np.array([[-1,-1,-1], [ 0, 0, 0], [ 1, 1, 1]], dtype=np.float32)
    # Manual 2D convolution (valid padding)
    from scipy.ndimage import convolve
    gx = convolve(gray_img.astype(np.float32), kx)
    gy = convolve(gray_img.astype(np.float32), ky)
    magnitude = np.sqrt(gx**2 + gy**2)
    return np.clip(magnitude, 0, 255).astype(np.uint8)

def manual_laplacian(gray_img):
    """4-neighbour Laplacian kernel via NumPy."""
    kernel = np.array([[0,  1, 0],
                       [1, -4, 1],
                       [0,  1, 0]], dtype=np.float32)
    from scipy.ndimage import convolve
    lap = convolve(gray_img.astype(np.float32), kernel)
    lap = np.abs(lap)
    return np.clip(lap, 0, 255).astype(np.uint8)

def cv2_sobel(gray_img):
    """cv2 Sobel — used as reference baseline."""
    sx = cv2.Sobel(gray_img, cv2.CV_64F, 1, 0, ksize=3)
    sy = cv2.Sobel(gray_img, cv2.CV_64F, 0, 1, ksize=3)
    mag = np.sqrt(sx**2 + sy**2)
    return np.clip(mag, 0, 255).astype(np.uint8)

# Apply on sample grayscale image
prewitt = manual_prewitt(gray)
laplacian = manual_laplacian(gray)
sobel_cv2 = cv2_sobel(gray)

fig, axes = plt.subplots(1, 4, figsize=(16, 4))
fig.suptitle("Edge Detection Comparison", fontsize=13, fontweight='bold')
for ax, img, title in zip(axes,
    [gray, prewitt, laplacian, sobel_cv2],
    ["Original Gray", "Manual Prewitt", "Manual Laplacian", "cv2 Sobel"]):
    ax.imshow(img, cmap='gray')
    ax.set_title(title, fontsize=10)
    ax.axis('off')
plt.tight_layout()
plt.savefig("edge_detection.png", dpi=100)
plt.show()
```

```python
# ────────────────────────────────────────────────────────────
# CELL 6 — Build Feature Vector for Shallow NN
# ────────────────────────────────────────────────────────────

def extract_features(img_bgr):
    """
    Combine normalised pixel values + Sobel edge magnitudes
    into a flat feature vector for the shallow NN.
    """
    gray = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2GRAY)
    norm = gray.astype(np.float32) / 255.0
    edges = cv2_sobel(gray).astype(np.float32) / 255.0
    return np.concatenate([norm.flatten(), edges.flatten()])

print("Extracting features for all images …")
X_feat = np.array([extract_features(img) for img in X_raw])
print(f"Feature matrix shape: {X_feat.shape}")

X_tr_f, X_val_f, y_tr_f, y_val_f = train_test_split(
    X_feat, y, test_size=0.2, random_state=42, stratify=y
)

```

```python
# ────────────────────────────────────────────────────────────
# CELL 7 — Shallow Fully-Connected NN (Baseline)
# ────────────────────────────────────────────────────────────

print("\nTraining shallow FC NN (sklearn MLP) …")
mlp = MLPClassifier(
    hidden_layer_sizes=(256,),
    activation='relu',
    max_iter=30,
    random_state=42,
    verbose=True
)
mlp.fit(X_tr_f, y_tr_f)
baseline_acc = accuracy_score(y_val_f, mlp.predict(X_val_f))
print(f"\n✅ Shallow NN Validation Accuracy: {baseline_acc:.4f} ({baseline_acc*100:.2f}%)")

# Plot loss curve
plt.figure(figsize=(8, 4))
plt.plot(mlp.loss_curve_, color='steelblue', linewidth=2)
plt.title("Shallow FC NN — Training Loss Curve", fontsize=13, fontweight='bold')
plt.xlabel("Iteration")
plt.ylabel("Loss")
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.savefig("shallow_nn_loss.png", dpi=100)
plt.show()
```

```python
# ────────────────────────────────────────────────────────────
# CELL 8 — Prepare Data for CNN (Keras)
# ────────────────────────────────────────────────────────────

# Normalize pixel values & one-hot encode labels
X_norm = np.array([classical_preprocess(img)[3] for img in X_raw])   # CLAHE-enhanced float [0,1]
y_cat  = to_categorical(y, N_CLASSES)

X_train, X_val, y_train, y_val = train_test_split(
    X_norm, y_cat, test_size=0.2, random_state=42, stratify=y
)
print(f"Train: {X_train.shape} | Val: {X_val.shape}")

# Data augmentation for training
datagen = ImageDataGenerator(
    rotation_range=10,
    width_shift_range=0.1,
    height_shift_range=0.1,
    zoom_range=0.1,
    horizontal_flip=False,   # traffic signs are NOT horizontally symmetric
)
datagen.fit(X_train)
```

```python
# ────────────────────────────────────────────────────────────
# CELL 9 — CNN from Scratch + Hyperparameter Experiments
#   Experiment 1: Learning rate  (0.001 vs 0.0001)
#   Experiment 2: Batch size     (32 vs 128)
# ────────────────────────────────────────────────────────────

def build_cnn(input_shape=(32, 32, 3), n_classes=43, lr=0.001):
    """
    Custom CNN:
      Conv(32) → BN → Conv(32) → BN → MaxPool → Dropout(0.25)
      Conv(64) → BN → Conv(64) → BN → MaxPool → Dropout(0.25)
      Flatten → Dense(256) → BN → Dropout(0.5) → Dense(n_classes)
    """
    model = keras.Sequential([
        # Block 1
        layers.Conv2D(32, (3,3), padding='same', activation='relu', input_shape=input_shape),
        layers.BatchNormalization(),
        layers.Conv2D(32, (3,3), padding='same', activation='relu'),
        layers.BatchNormalization(),
        layers.MaxPooling2D((2,2)),
        layers.Dropout(0.25),
        # Block 2
        layers.Conv2D(64, (3,3), padding='same', activation='relu'),
        layers.BatchNormalization(),
        layers.Conv2D(64, (3,3), padding='same', activation='relu'),
        layers.BatchNormalization(),
        layers.MaxPooling2D((2,2)),
        layers.Dropout(0.25),
        # Head
        layers.Flatten(),
        layers.Dense(256, activation='relu'),
        layers.BatchNormalization(),
        layers.Dropout(0.5),
        layers.Dense(n_classes, activation='softmax'),
    ], name="CustomCNN")

    model.compile(
        optimizer=keras.optimizers.Adam(learning_rate=lr),
        loss='categorical_crossentropy',
        metrics=['accuracy']
    )
    return model

EPOCHS = 1

def run_experiment(lr, batch_size, label):
    """Train CNN and return history dict."""
    print(f"\n{'='*55}")
    print(f"  Experiment: {label}  (lr={lr}, batch={batch_size})")
    print(f"{'='*55}")
    model = build_cnn(lr=lr)
    history = model.fit(
        datagen.flow(X_train, y_train, batch_size=batch_size),
        steps_per_epoch=len(X_train) // batch_size,
        validation_data=(X_val, y_val),
        epochs=EPOCHS,
        verbose=1
    )
    val_acc = max(history.history['val_accuracy'])
    print(f"  Best Val Accuracy: {val_acc:.4f}")
    return history, model, val_acc

# Run all 4 configurations
history_lr001_bs32,  model_a, acc_a = run_experiment(lr=0.001,  batch_size=32,  label="LR=0.001, BS=32")
history_lr0001_bs32, model_b, acc_b = run_experiment(lr=0.0001, batch_size=32,  label="LR=0.0001, BS=32")
history_lr001_bs128, model_c, acc_c = run_experiment(lr=0.001,  batch_size=128, label="LR=0.001, BS=128")
history_lr0001_bs128,model_d, acc_d = run_experiment(lr=0.0001, batch_size=128, label="LR=0.0001, BS=128")


# ────────────────────────────────────────────────────────────
# CELL 10 — Plot Training vs Validation Curves (all configs)
# ────────────────────────────────────────────────────────────

configs = {
    "LR=0.001 BS=32":   history_lr001_bs32,
    "LR=0.0001 BS=32":  history_lr0001_bs32,
    "LR=0.001 BS=128":  history_lr001_bs128,
    "LR=0.0001 BS=128": history_lr0001_bs128,
}
colors = ['steelblue', 'tomato', 'seagreen', 'darkorange']

fig, axes = plt.subplots(2, 2, figsize=(14, 10))
fig.suptitle("CNN Hyperparameter Experiments — Train vs Val Curves",
             fontsize=14, fontweight='bold')

for ax, (label, hist), color in zip(axes.flat, configs.items(), colors):
    ep = range(1, EPOCHS + 1)
    ax.plot(ep, hist.history['accuracy'],     color=color,  linewidth=2,       label='Train Acc')
    ax.plot(ep, hist.history['val_accuracy'], color=color,  linewidth=2, ls='--', label='Val Acc')
    ax.plot(ep, hist.history['loss'],         color='gray', linewidth=1,       label='Train Loss', alpha=0.6)
    ax.plot(ep, hist.history['val_loss'],     color='gray', linewidth=1, ls='--', label='Val Loss', alpha=0.6)
    ax.set_title(label, fontsize=11)
    ax.set_xlabel("Epoch")
    ax.set_ylabel("Accuracy / Loss")
    ax.legend(fontsize=8)
    ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig("cnn_hyperparameter_curves.png", dpi=100)
plt.show()

# Summary table
print("\n📊 Hyperparameter Experiment Summary")
print(f"{'Configuration':<25} {'Best Val Acc':>12}")
print("-" * 40)
for (label, _), acc in zip(configs.items(), [acc_a, acc_b, acc_c, acc_d]):
    print(f"{label:<25} {acc*100:>10.2f}%")
```

```python
# ────────────────────────────────────────────────────────────
# CELL 11 — Transfer Learning: ResNet50 & EfficientNetB0
#   • Use as frozen feature extractors
#   • Benchmark inference time
# ────────────────────────────────────────────────────────────

# GTSRB is 32×32; pretrained models need ≥75×75.
# We upsample to 96×96 for compatibility.
TL_SIZE = 96

X_tl = np.array([cv2.resize(img, (TL_SIZE, TL_SIZE)) for img in X_raw], dtype=np.float32) / 255.0
X_tr_tl, X_val_tl, y_tr_tl, y_val_tl = train_test_split(
    X_tl, y_cat, test_size=0.2, random_state=42, stratify=y
)

def build_transfer_model(base_fn, input_shape, n_classes=43, name="TL"):
    """Frozen backbone + custom classification head."""
    base = base_fn(include_top=False, weights='imagenet', input_shape=input_shape)
    base.trainable = False   # freeze all base layers

    inputs  = keras.Input(shape=input_shape)
    x       = base(inputs, training=False)
    x       = layers.GlobalAveragePooling2D()(x)
    x       = layers.Dropout(0.3)(x)
    outputs = layers.Dense(n_classes, activation='softmax')(x)
    model   = keras.Model(inputs, outputs, name=name)
    model.compile(
        optimizer=keras.optimizers.Adam(1e-3),
        loss='categorical_crossentropy',
        metrics=['accuracy']
    )
    return model, base

print("\nBuilding ResNet50 feature extractor …")
resnet_model, resnet_base = build_transfer_model(
    ResNet50, (TL_SIZE, TL_SIZE, 3), name="ResNet50_FE"
)
print(f"ResNet50 total params:     {resnet_model.count_params():,}")
print(f"ResNet50 trainable params: {sum([np.prod(v.shape) for v in resnet_model.trainable_variables]):,}")

print("\nBuilding EfficientNetB0 feature extractor …")
effnet_model, effnet_base = build_transfer_model(
    EfficientNetB0, (TL_SIZE, TL_SIZE, 3), name="EfficientNetB0_FE"
)
print(f"EfficientNetB0 total params:     {effnet_model.count_params():,}")
print(f"EfficientNetB0 trainable params: {sum([np.prod(v.shape) for v in effnet_model.trainable_variables]):,}")

TL_EPOCHS = 2

print("\nTraining ResNet50 head …")
hist_resnet = resnet_model.fit(
    X_tr_tl, y_tr_tl,
    batch_size=64,
    epochs=TL_EPOCHS,
    validation_data=(X_val_tl, y_val_tl),
    verbose=1
)

print("\nTraining EfficientNetB0 head …")
hist_effnet = effnet_model.fit(
    X_tr_tl, y_tr_tl,
    batch_size=64,
    epochs=TL_EPOCHS,
    validation_data=(X_val_tl, y_val_tl),
    verbose=1
)
```

```python
# ────────────────────────────────────────────────────────────
# CELL 12 — Benchmark Inference Time
# ────────────────────────────────────────────────────────────

def measure_inference_time(model, x_batch, n_runs=20):
    """Average inference time over n_runs forward passes."""
    _ = model.predict(x_batch[:1], verbose=0)   # warm-up
    times = []
    for _ in range(n_runs):
        t0 = time.perf_counter()
        model.predict(x_batch[:32], verbose=0)
        times.append((time.perf_counter() - t0) / 32 * 1000)   # ms/image
    return np.mean(times), np.std(times)

# Use best CNN config (model_a: LR=0.001, BS=32)
x_bench_cnn = X_val[:32]
x_bench_tl  = X_val_tl[:32]

t_cnn_mean,  t_cnn_std  = measure_inference_time(model_a,       x_bench_cnn)
t_rn50_mean, t_rn50_std = measure_inference_time(resnet_model,  x_bench_tl)
t_eff_mean,  t_eff_std  = measure_inference_time(effnet_model,  x_bench_tl)

acc_cnn  = max(history_lr001_bs32.history['val_accuracy'])
acc_rn50 = max(hist_resnet.history['val_accuracy'])
acc_eff  = max(hist_effnet.history['val_accuracy'])

param_cnn  = model_a.count_params()
param_rn50 = resnet_model.count_params()
param_eff  = effnet_model.count_params()

print("\n📊 Model Comparison Table")
print(f"{'Model':<20} {'Val Acc':>8} {'Params':>12} {'ms/img':>10}")
print("-" * 55)
for name, acc, params, t_mean in [
    ("Custom CNN",       acc_cnn,  param_cnn,  t_cnn_mean),
    ("ResNet50-FE",      acc_rn50, param_rn50, t_rn50_mean),
    ("EfficientNetB0-FE",acc_eff,  param_eff,  t_eff_mean),
]:
    print(f"{name:<20} {acc*100:>7.2f}% {params:>12,} {t_mean:>9.3f}")
```

```python
# ─────────────────────────────────────────────────────────────
# Manual Prewitt kernels implemented with NumPy
# ─────────────────────────────────────────────────────────────

# Prewitt kernels (3×3)
PREWITT_KX = np.array([[-1,  0,  1],
                        [-1,  0,  1],
                        [-1,  0,  1]], dtype=np.float32)

PREWITT_KY = np.array([[-1, -1, -1],
                        [ 0,  0,  0],
                        [ 1,  1,  1]], dtype=np.float32)


def apply_prewitt_manual(gray_img_uint8):
    """
    Apply Prewitt edge detector using manual NumPy convolution via cv2.filter2D.
    Input: uint8 grayscale image.
    Returns: magnitude image (uint8).
    """
    gx = cv2.filter2D(gray_img_uint8.astype(np.float32), -1, PREWITT_KX)
    gy = cv2.filter2D(gray_img_uint8.astype(np.float32), -1, PREWITT_KY)
    magnitude = np.sqrt(gx**2 + gy**2)
    magnitude = np.clip(magnitude, 0, 255).astype(np.uint8)
    return magnitude


def apply_sobel_cv2(gray_img_uint8):
    """
    Apply Sobel edge detector using cv2.Sobel (built-in).
    Returns: magnitude image (uint8).
    """
    sobelx = cv2.Sobel(gray_img_uint8, cv2.CV_64F, 1, 0, ksize=3)
    sobely = cv2.Sobel(gray_img_uint8, cv2.CV_64F, 0, 1, ksize=3)
    magnitude = np.sqrt(sobelx**2 + sobely**2)
    magnitude = np.clip(magnitude, 0, 255).astype(np.uint8)
    return magnitude


def apply_laplacian(gray_img_uint8):
    """
    Apply Laplacian of Gaussian edge detector.
    Returns: uint8 absolute result.
    """
    blurred = cv2.GaussianBlur(gray_img_uint8, (3, 3), 0)
    lap = cv2.Laplacian(blurred, cv2.CV_64F, ksize=3)
    lap_abs = np.clip(np.abs(lap), 0, 255).astype(np.uint8)
    return lap_abs


# Apply all three filters on the sample image
gray_uint8 = (gray_n * 255).astype(np.uint8)

prewitt_out  = apply_prewitt_manual(gray_uint8)
sobel_out    = apply_sobel_cv2(gray_uint8)
laplacian_out = apply_laplacian(gray_uint8)

fig, axes = plt.subplots(1, 4, figsize=(16, 4))
axes[0].imshow(gray_uint8,    cmap='gray'); axes[0].set_title('Grayscale Input')
axes[1].imshow(prewitt_out,   cmap='gray'); axes[1].set_title('Prewitt (Manual Kernels)')
axes[2].imshow(sobel_out,     cmap='gray'); axes[2].set_title('Sobel (cv2.Sobel)')
axes[3].imshow(laplacian_out, cmap='gray'); axes[3].set_title('Laplacian (LoG)')
for ax in axes: ax.axis('off')
plt.suptitle('Edge Feature Extraction Comparison', fontsize=13)
plt.tight_layout()
plt.show()

# Pixel-wise difference between Prewitt and Sobel
diff = np.abs(prewitt_out.astype(int) - sobel_out.astype(int))
print(f"Mean pixel difference (Prewitt vs Sobel): {diff.mean():.4f}")
print("Both detect similar edges; Sobel is slightly smoother due to Gaussian weighting.")
```
